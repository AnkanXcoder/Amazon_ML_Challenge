## Objective

Design a preprocessing strategy for the business entity resolution datasets before creating ML-ready features.

The preprocessing should:

- Preserve the original raw data.
- Reduce formatting-based differences.
- Handle missing values safely.
- Keep preprocessing consistent across Source 1, Source 2, and Source 3.
- Avoid aggressive transformations that could create false matches.

---

## 1. Business Name

| Problem | Preprocessing | Reason |
|---|---|---|
| Upper/lowercase differences | Convert to lowercase | Make equivalent text comparable |
| Punctuation differences | Remove punctuation | Reduce formatting variation |
| Extra whitespace | Normalize whitespace | Ensure consistent spacing |
| Unicode variations | Unicode normalization | Handle different Unicode representations |
| Missing values | Fill with empty string + missing indicator | Prevent processing errors while preserving missingness |
| Abbreviations | Do not automatically expand | Avoid incorrect transformations |
| Typos | Do not automatically correct | Avoid introducing false matches |
| Legal suffixes | Do not automatically remove | Preserve potentially useful information |

---

## 2. Business Address

| Problem | Preprocessing | Reason |
|---|---|---|
| Upper/lowercase differences | Convert to lowercase | Make addresses comparable |
| Punctuation differences | Remove punctuation | Reduce formatting variation |
| Extra whitespace | Normalize whitespace | Ensure consistent formatting |
| Unicode variations | Unicode normalization | Handle multilingual/Unicode representations |
| Missing addresses | Fill with empty string + missing indicator | Preserve missingness information |
| Address abbreviations | Do not automatically expand initially | Avoid incorrect transformations |
| Address components in different order | Do not reorder initially | Preserve original information |

---

## 3. Country

| Problem | Preprocessing | Reason |
|---|---|---|
| Case differences | Convert to lowercase | Consistent categorical representation |
| Extra whitespace | Strip whitespace | Remove formatting differences |
| Missing values | No imputation required | No missing country values were found |

---

## 4. Transformations We Will Not Apply Automatically

The initial preprocessing will **not** automatically:

- Correct spelling mistakes
- Expand abbreviations such as `Cir` → `Circle`
- Expand abbreviations such as `Rd` → `Road`
- Remove legal suffixes such as `LLC`, `Inc`, or `Pvt Ltd`
- Translate business names or addresses
- Create manual aliases
- Reorder address components

These transformations will only be considered later if analysis of the training data provides sufficient evidence.

## Basic Text Preprocessing

````python
import re
import unicodedata

def normalize_text(text):
    if pd.isna(text):  # Check for missing values
        return ""

    text = str(text).lower()
    text = unicodedata.normalize("NFKC", text) # Different Unicode representations → standardized representation
    text = re.sub(r"[^\w\s]", " ", text) # Remove punctuation and special characters
    text = re.sub(r"\s+", " ", text).strip() # remove extra space

    return text
````

````python
s1["business_name_clean"] = s1["business_name"].map(normalize_text)

s1["business_address_clean"] = s1["business_address"].map(normalize_text)

s1["country_clean"] = (
    s1["country"]
    .str.lower()
    .str.strip()
)
````

# Analyze Name Token Structure

````python
def name_token_count(text):
    if not text:
        return 0
    return len(text.split())


for name, df in [("S1", s1), ("S2", s2), ("S3", s3)]:

    token_counts = df["business_name_clean"].map(name_token_count)

    print("=" * 50)
    print(name)

    print("Average tokens:", token_counts.mean())
    print("Minimum tokens:", token_counts.min())
    print("Maximum tokens:", token_counts.max())

    print("\nToken distribution:")
    print(token_counts.value_counts().sort_index().head(15))
````

- Most business names contain 3–4 tokens.
- S2 and S3 contain a small number of names with 0 tokens, corresponding to the missing names we already identified.
- S2/S3 have considerably longer names than the typical S1 name.
Token count can therefore become a useful ML feature, but we should not use it as a matching rule by itself.

# Basic Name Features

````python
def add_name_features(df):

    df["name_char_length"] = (
        df["business_name_clean"].str.len().astype("int32")
    )

    df["name_token_count"] = (
        df["business_name_clean"]
        .str.split()
        .str.len()
        .astype("int16")
    )

    df["name_unique_token_count"] = (
        df["business_name_clean"]
        .str.split()
        .map(lambda x: len(set(x)))
        .astype("int16")
    )

    return df
````

````python
s1 = add_name_features(s1)
s2 = add_name_features(s2)
s3 = add_name_features(s3)
````

````python
display(
    s1[
        [
            "business_name_clean",
            "name_char_length",
            "name_token_count",
            "name_unique_token_count"
        ]
    ].head(10)
)
````

# Basic Address Features

````python
def add_address_features(df):

    df["address_char_length"] = (
        df["business_address_clean"].str.len().astype("int32")
    )

    df["address_token_count"] = (
        df["business_address_clean"]
        .str.split()
        .str.len()
        .astype("int16")
    )

    df["address_digit_count"] = (
        df["business_address_clean"]
        .str.count(r"\d")
        .astype("int16")
    )

    df["address_unique_token_count"] = (
        df["business_address_clean"]
        .str.split()
        .map(lambda x: len(set(x)))
        .astype("int16")
    )

    return df
````

````python
s1 = add_address_features(s1)
s2 = add_address_features(s2)
s3 = add_address_features(s3)
````
